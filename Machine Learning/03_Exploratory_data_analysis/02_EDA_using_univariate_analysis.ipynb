# EDA - to understand data in and out
- Univariate analysis: single variable analysis (independent columns)
analysis depends on what type of data is it: numerical(height/age...) or categorical (placed/marital status...)

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('/content/train.csv')
```

```python
df.head()
```

# 1. Categorical Data

a. Countplot: Frequency count of all the categories (similar to pandas value_counts)

```python
sns.countplot(x='Survived', data=df)
df['Survived'].value_counts()
```

b. Piechart

```python
sns.set_style('darkgrid')
df['Survived'].value_counts().plot(kind='pie', autopct='%.2f')
```

# 2. Numerical data
a. Histogram

```python
df['Age'].value_counts().plot(kind='hist',bins=50)
```

b. Distplot

```python
df['Age'].value_counts().plot(kind='kde')
```

```python
sns.distplot(df['Age'])
```

c. Boxplot

```python
sns.boxplot(x='Survived', y='Age', data=df)
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

Webpage fetching

```python
url = "http://quotes.toscrape.com/"
response = requests.get(url)
print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

Print the html code

```python
soup
```

Finding all the quote blocks

```python
quotes = soup.find_all("div", class_="quote")
print(len(quotes))
```

Extracting a quote

```python
quote = quotes[0]

text = quote.find("span", class_="text").text
author = quote.find("small", class_="author").text

print(text+'---'+author)
```

Search for the tags

```python
tags = quote.find_all("a", class_="tag")
for tag in tags:
    print(tag.text)
```

Extracting everything from a page

```python
data = []
for quote in quotes:
    text = quote.find("span", class_="text").text
    author = quote.find("small", class_="author").text
    tags = [
        tag.text
        for tag in quote.find_all("a", class_="tag")
    ]
    data.append({
        "Quote": text,
        "Author": author,
        "Tags": ", ".join(tags)
    })
```

```python
df = pd.DataFrame(data)
df
```

```python
df.to_csv("quotes.csv", index=False)
```
